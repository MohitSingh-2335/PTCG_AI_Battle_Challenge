```python
import pandas as pd

# Load the dataset
csv_path = r'Data\EN_Card_Data.csv'
df = pd.read_csv(csv_path)

# Show the first 5 rows in a beautiful table
df.head()
```

```python
print(f"Total number of cards: {len(df)}")
print(f"Columns: {list(df.columns)}")

# Count Card Categories
df['Category'].value_counts()
```

```python
# Filter to only Pokémon cards
pokemon_cards = df[df['Category'] == 'Pokémon']

# Show the top 5 most common Elemental Types
pokemon_cards['Type'].value_counts().head(5)
```

```python
# Let's see what the main types of cards actually are!
main_column = 'Stage (Pokémon)/Type (Energy and Trainer)'

print("--- Main Card Types ---")
df[main_column].value_counts()
```

```python
# 1. Select only the actual creature cards
main_column = 'Stage (Pokémon)/Type (Energy and Trainer)'
creature_types = ['Basic Pokémon', 'Stage 1 Pokémon', 'Stage 2 Pokémon']

# We use .copy() so we can safely edit the new mini-spreadsheet without warnings
pokemon_cards = df[df[main_column].isin(creature_types)].copy() 

# 2. Show the top 5 most common Elemental Types
print("--- Top 5 Pokémon Types ---")
print(pokemon_cards['Type'].value_counts().head(5))

# 3. Analyze HP!
# The HP column might be read as text by default, so we force it to be a number
pokemon_cards['HP'] = pd.to_numeric(pokemon_cards['HP'], errors='coerce')

print("\n--- HP Stats ---")
print(f"Average HP: {pokemon_cards['HP'].mean():.1f}")
print(f"Highest HP: {pokemon_cards['HP'].max()}")

# Let's find out WHO has the highest HP!
highest_hp_pokemon = pokemon_cards[pokemon_cards['HP'] == pokemon_cards['HP'].max()]
print("\nPokémon with the highest HP:")
highest_hp_pokemon[['Card Name', 'HP', 'Type']]
```
